## **Praktikum 1**

**Menyiapkan Library**

```python
import numpy as np
```

**Langkah 1. Membuat Dataset Sederhana (XOR)**

```python
# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])
```

**Langkah 2. Inisialisasi Parameter**

```python
# Parameter
input_size = 2
hidden_size = 2
output_size = 1
lr = 0.1
```

**Langkah 3. Inisialisasi Bobot dan Bias**

```python
# Inisialisasi bobot
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))
```

**Langkah 4. Membuat Fungsi Aktivasi**

```python
# Fungsi aktivasi
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)
```

**Langkah 5. Melakukan Training ANN**

```python
# Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")
```

**Langkah 6. Menampilkan Output Akhir**

```python
# Output akhir
print("Prediksi:")
print(a2)
```

**Tugas 1.3 – Menambahkan Fungsi Aktivasi ReLU dan Membandingkan Hasil**

Membuat Fungsi Aktivasi ReLU

```python
# Fungsi aktivasi ReLU
def relu(x):
    return np.maximum(0, x)

# Turunan fungsi aktivasi ReLU
def relu_derivative(x):
    return (x > 0).astype(float)
```

Inisialisasi Parameter dan Bobot

```python
# Parameter
input_size = 2
hidden_size = 3
output_size = 1
lr = 0.1

# Inisialisasi bobot
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))
```

Implementasi Training ANN dengan ReLU

```python
# Training menggunakan ReLU
loss_relu = []

for epoch in range(10000):

    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = relu(z1)

    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * relu_derivative(z1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        loss_relu.append(loss)
        print(f"Epoch {epoch}, Loss ReLU: {loss}")
```

Analisis Perbandingan Sigmoid dan ReLU

Berdasarkan hasil pengujian ANN, fungsi aktivasi Sigmoid dan ReLU menghasilkan nilai Loss yang berbeda selama proses training. Sigmoid menghasilkan keluaran antara 0 hingga 1, sedangkan ReLU menghasilkan nilai 0 untuk input negatif dan mempertahankan nilai positif. Perbedaan fungsi aktivasi tersebut memengaruhi proses pembelajaran dan hasil prediksi model.

Fungsi aktivasi dengan nilai Loss lebih kecil menunjukkan hasil pelatihan yang lebih baik berdasarkan metrik MSE pada pengujian ini. Oleh karena itu, pemilihan fungsi aktivasi perlu mempertimbangkan hasil evaluasi model.

## **Tugas 4 - Mengubah Learning Rate**

**Percobaan Perubahan Learning Rate & Perbandingan Hasil Loss**

```python
# Daftar learning rate
learning_rates = [0.1, 0.01, 0.001]

hasil_loss = []

for lr in learning_rates:

    # Membuat model
    model = tf.keras.Sequential([
        tf.keras.layers.Dense(
            10,
            activation='relu',
            input_shape=(1,)
        ),
        tf.keras.layers.Dense(1)
    ])

    # Compile model
    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=lr
        ),
        loss='mse'
    )

    # Training
    history = model.fit(
        X_train,
        y_train,
        epochs=100,
        verbose=0
    )

    # Menyimpan loss terakhir
    loss_akhir = history.history['loss'][-1]

    hasil_loss.append({
        'Learning Rate': lr,
        'Loss': loss_akhir
    })

# Menampilkan hasil
hasil_perbandingan = pd.DataFrame(hasil_loss)
hasil_perbandingan
```

**Analisis Hasil:**

Berdasarkan percobaan, perbedaan learning rate memengaruhi kecepatan dan kestabilan proses pelatihan model ANN. Learning rate 0.1 memiliki langkah pembaruan bobot lebih besar, 0.01 memberikan langkah yang lebih moderat, sedangkan 0.001 menghasilkan pembaruan lebih kecil dan bertahap. Learning rate terbaik ditentukan berdasarkan nilai loss terkecil dan kestabilan proses pelatihan.


**Kesimpulan:**

Pemilihan learning rate yang tepat diperlukan agar model ANN dapat mencapai hasil pelatihan yang optimal dan stabil.

## **Praktikum Lanjutan – MLP Regresi Menggunakan Dataset Boston**

**1. Import library**

```python
# MLP regresi (Keras)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
```

**2. Memuat dataset Boston**

```python
# 1. Load

data_url = "http://lib.stat.cmu.edu/datasets/boston"

raw_df = pd.read_csv(
    data_url,
    sep=r"\s+",
    skiprows=22,
    header=None
)

X = np.hstack([
    raw_df.values[::2, :],
    raw_df.values[1::2, :2]
])

y = raw_df.values[1::2, 2]
```

**Periksa Dataset:**

```python
print("Ukuran X:", X.shape)
print("Ukuran y:", y.shape)
print("\nLima data pertama:")
print(X[:5])

print("\nLima target pertama:")
print(y[:5])
```

**3. Preprocessing data**

```python
# 2. Preprocess

scaler = StandardScaler()

Xs = scaler.fit_transform(X)

X_train, X_val, y_train, y_val = train_test_split(
    Xs,
    y,
    test_size=0.2,
    random_state=42
)
```

**Periksa Hasil:**

```python
print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("y_train:", y_train.shape)
print("y_val:", y_val.shape)
```

**4. Membangun model ANN**

```python
# 3. Build model

model = Sequential([
    Dense(
        64,
        activation='relu',
        input_shape=(X_train.shape[1],)
    ),
    Dense(32, activation='relu'),
    Dense(1)
])
```

**5. Compile model**

```python
model.compile(
    optimizer=Adam(1e-3),
    loss='mse',
    metrics=['mae']
)
```

**6. Melatih model**

```python
# 4. Train

h = model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=200,
    batch_size=32,
    verbose=0
)
```

**7. Visualisasi hasil pelatihan**

```python
# 5. Plot

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)

plt.plot(
    h.history['loss'],
    label='train_loss'
)

plt.plot(
    h.history['val_loss'],
    label='val_loss'
)

plt.legend()
plt.title('MSE')

plt.subplot(1, 2, 2)

plt.plot(
    h.history['mae'],
    label='train_mae'
)

plt.plot(
    h.history['val_mae'],
    label='val_mae'
)

plt.legend()
plt.title('MAE')

plt.show()
```

**8. Evaluasi model menggunakan RMSE**

```python
from sklearn.metrics import mean_squared_error

pred = model.predict(X_val)

print(
    'RMSE:',
    np.sqrt(mean_squared_error(y_val, pred))
)
```

**10 – Membuat fungsi eksperimen**

```python
def eksperimen_ann(neuron, aktivasi='relu', epochs=10):

    model = Sequential([
        Flatten(input_shape=(28, 28)),
        Dense(neuron[0], activation=aktivasi),
        Dense(neuron[1], activation=aktivasi),
        Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    start = time.time()

    history = model.fit(
        X_train,
        y_train,
        epochs=epochs,
        batch_size=32,
        validation_split=0.2,
        verbose=0
    )

    waktu = time.time() - start

    loss, akurasi = model.evaluate(
        X_test,
        y_test,
        verbose=0
    )

    return akurasi, waktu, loss
```

**11 – Percobaan perubahan jumlah neuron**

```python
hasil_neuron = []

for neuron in [(128, 64), (256, 128)]:

    acc, waktu, loss = eksperimen_ann(neuron)

    hasil_neuron.append({
        'Arsitektur': str(neuron),
        'Akurasi': acc,
        'Loss': loss,
        'Waktu Pelatihan': waktu
    })

df_neuron = pd.DataFrame(hasil_neuron)

df_neuron
```

**12 – Menambahkan satu hidden layer**

```python
model_3_layer = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(10, activation='softmax')
])

model_3_layer.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

start = time.time()

history_3_layer = model_3_layer.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2,
    verbose=0
)

waktu_3_layer = time.time() - start

loss_3_layer, acc_3_layer = model_3_layer.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("Akurasi:", acc_3_layer)
print("Loss:", loss_3_layer)
print("Waktu pelatihan:", waktu_3_layer)
```
